# Bug Evolution Lab — Interactive Playback

This notebook is a thin client over `buglab.playback` and `buglab.notebook_ui`. Optimization runs only when **Run / load** is pressed. Moving sliders and changing display toggles replay recorded genomes without rerunning the optimizer.

```python
from pathlib import Path
from IPython.display import display
from buglab.notebook_ui import create_playback_widget
from buglab.playback import ExperimentCache

cache = ExperimentCache()
example = Path('examples/task4_demo.npz')
if example.exists():
    cache.load(example)
ui = create_playback_widget(cache, output_dir=Path('outputs_notebook'))
ui.display() if hasattr(ui, 'display') else display(ui)
```

## Act I Demo

Recommended minimal presentation sequence:

1. Select the cached GA run and generation 0. Keep **All trajectories**, **Selected individual**, and **Best-so-far** enabled to introduce a population of turn-angle programs.
2. Press Play. Point out that the horizontal progress axes are fitness **evaluations**, while the map advances by recorded generations.
3. Pause on a later generation, enable **Collision markers**, and inspect the selected genome's angle preview and physical outcome metrics.
4. Choose `random` or `local`, keep the same map, fitness, seed, population, and batches, then press **Run / load** explicitly. Random search shows fresh independent candidates; local search shows its incumbent and neighbors.
5. Use **Export generation PNG** and **Export summary PNG** for white-background presentation slides.

Raw fitness is meaningful only within one fixed fitness strategy. Across different strategies, use goal success, steps, collisions, path length, final distance, and evaluation budget.

## Controls and caching

Changing map, method, fitness, seed, or budget marks the settings as pending; it never relabels the displayed run. Press **Run / load** to execute a new configuration or retrieve an exact cached match. The generation slider, Play control, individual selector, trajectory toggles, and top-k filter only update playback.

## Static fallback

If widgets are unavailable, install `requirements-notebook.txt`. The returned fallback object can still render static figures with `render_static(data, snapshot_index)`, and `buglab.playback.export_generation_png` / `export_summary_png` remain usable without widgets.
